Section 1: Resume Information Extraction

```python
import re
resume = """
Name: Shaik Mubarak
Email: mubarak@gmail.com
Mobile: 9876543210
Skills: Python, Java, SQL, Machine Learning, NLP
Experience: 3 years
"""
name = re.search(r"Name:\s*(.*)", resume)
email = re.search(r"[\w.-]+@[\w.-]+\.\w+", resume)
mobile = re.search(r"\b\d{10}\b", resume)
experience = re.search(r"(\d+)\s+years?", resume)
skills = ["Python", "Java", "SQL", "Machine Learning", "NLP"]
found_skills = []
for skill in skills:
    if re.search(r"\b" + re.escape(skill) + r"\b", resume, re.I):
        found_skills.append(skill)
print("----- CANDIDATE PROFILE -----")
print("Name:", name.group(1))
print("Email:", email.group())
print("Mobile:", mobile.group())
print("Skills:", ", ".join(found_skills))
print("Experience:", experience.group(1), "years")
years = int(experience.group(1))
if years >= 2 and "Python" in found_skills:
    print("Eligibility: ELIGIBLE")
else:
    print("Eligibility: NOT ELIGIBLE")
```

Section 2: Product Search System

```python
import re
products = [
    "Python Programming Book",
    "Java Programming Book",
    "Python Laptop",
    "Machine Learning Course",
    "NLP Course",
    "SQL Database",
    "Python Notebook"
]
def search_product(keyword, search_type):
    result = []
    for product in products:
        if search_type == "exact":
            pattern = r"\b" + re.escape(keyword) + r"\b"

        elif search_type == "prefix":
            pattern = r"\b" + re.escape(keyword)

        elif search_type == "suffix":
            pattern = re.escape(keyword) + r"\b"
        elif search_type == "partial":
            pattern = re.escape(keyword)
        if re.search(pattern, product, re.I):
            result.append(product)
    print("\nSearch:", keyword)
    print("Type:", search_type)
    if result:
        print("Matching Products:")
        for product in result:
            print("-", product)
    else:
        print("No products found")
    print("Total Matches:", len(result))
search_product("Python", "exact")
search_product("Python", "prefix")
search_product("Course", "suffix")
search_product("gram", "partial")
search_product("python", "exact")
```

Section 3: University Registration System

```python
import re
reg_no = input("Enter Register Number: ")
email = input("Enter Institutional Email: ")
course = input("Enter Course Code: ")
semester = input("Enter Semester: ")
mobile = input("Enter Mobile Number: ")
reg_pattern = r"^\d{9}$"
email_pattern = r"^[a-zA-Z0-9._%+-]+@gmail\.com$"
course_pattern = r"^[A-Z]{3}\d{4}$"
semester_pattern = r"^(ONE|TWO|THREE|FOUR|FIVE|SIX|SEVEN|EIGHT)$"
mobile_pattern = r"^[6-9]\d{9}$"
if re.match(reg_pattern, reg_no):
    print("Register Number: Valid")
else:
    print("Register Number: Invalid")
if re.match(email_pattern, email):
    print("Email: Valid")
else:
    print("Email: Invalid")
if re.match(course_pattern, course):
    print("Course Code: Valid")
else:
    print("Course Code: Invalid")
if re.match(semester_pattern, semester, re.I):
    print("Semester: Valid")
else:
    print("Semester: Invalid")
if re.match(mobile_pattern, mobile):
    print("Mobile Number: Valid")
else:
    print("Mobile Number: Invalid")
if (re.match(reg_pattern, reg_no) and
    re.match(email_pattern, email) and
    re.match(course_pattern, course) and
    re.match(semester_pattern, semester, re.I) and
    re.match(mobile_pattern, mobile)):
    print("\nRegistration Status: SUCCESSFUL")
else:
    print("\nRegistration Status: FAILED")
```
