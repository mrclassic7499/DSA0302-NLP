1. Regular Expressions – Match and Search Patterns

```python
import re

text = "My email is student@gmail.com and my phone is 9876543210."
email = re.search(r'\w+@\w+\.\w+', text)
phone = re.search(r'\d{10}', text)

if email:
    print("Email found:", email.group())
else:
    print("Email not found")

if phone:
    print("Phone number found:", phone.group())
else:
    print("Phone number not found")
```

2. Finite State Automaton – Strings Ending with ab

```python
def fsa(string):
    state = 0
    for ch in string:
        if state == 0:
            if ch == 'a':
                state = 1
            else:
                state = 0
        elif state == 1:
            if ch == 'b':
                state = 2
            elif ch == 'a':
                state = 1
            else:
                state = 0
        elif state == 2:
            if ch == 'a':
                state = 1
            else:
                state = 0
    return state == 2
s = input("Enter a string: ")
if fsa(s):
    print("ACCEPTED")
else:
    print("REJECTED")
```

4. Finite-State Machine for Morphological Parsing – Plural Nouns

```python
def plural(noun):
    if noun.endswith("s"):
        return noun + "es"
    elif noun.endswith("x") or noun.endswith("ch") or noun.endswith("sh"):
        return noun + "es"
    else:
        return noun + "s"


words = ["cat", "dog", "box", "bus", "brush"]

for word in words:
    print(word, "->", plural(word))
```

6.N-Gram Model for Text Generation (Bigram)

```python
import random
from collections import defaultdict
text = "I love machine learning. I love Python programming. I love machine learning."
words = text.lower().replace(".", "").split()
bigrams = defaultdict(list)
for i in range(len(words) - 1):
    bigrams[words[i]].append(words[i + 1])
current = "i"
generated = [current]
for i in range(10):
    if current not in bigrams:
        break
    current = random.choice(bigrams[current])
    generated.append(current)
print("Original Text:", text)
print("Generated Text:", " ".join(generated))
```

8.Stochastic Part-of-Speech Tagging

```python
from collections import defaultdict, Counter
training_data = [
    [("I", "PRON"), ("love", "VERB"), ("Python", "NOUN")],
    [("I", "PRON"), ("like", "VERB"), ("Python", "NOUN")],
    [("Students", "NOUN"), ("love", "VERB"), ("learning", "NOUN")],
    [("Python", "NOUN"), ("is", "VERB"), ("easy", "ADJ")],
    [("Students", "NOUN"), ("like", "VERB"), ("learning", "NOUN")]
]
word_tags = defaultdict(Counter)
for sentence in training_data:
    for word, tag in sentence:
        word_tags[word.lower()][tag] += 1
text = "I love Python"
print("STOCHASTIC POS TAGGING")
print("----------------------")

for word in text.split():
    key = word.lower()

    if key in word_tags:
        tag = word_tags[key].most_common(1)[0][0]
    else:
        tag = "UNKNOWN"

    print(word, "->", tag)
```

9.Rule-Based POS Tagging Using Regular Expressions

```python
import re
def pos_tag(word):
    word = word.lower()

    if re.fullmatch(r"\d+", word):
        return "NUMBER"

    elif re.fullmatch(r"\w+ing", word):
        return "VERB"

    elif re.fullmatch(r"\w+ly", word):
        return "ADVERB"

    elif re.fullmatch(r"\w+(ous|ful|able)", word):
        return "ADJECTIVE"

    elif re.fullmatch(r"\w+ed", word):
        return "VERB"

    elif re.fullmatch(r"\w+s", word):
        return "NOUN (PLURAL)"

    else:
        return "NOUN"
text = "Students are learning Python quickly"
print("RULE-BASED POS TAGGING")
print("----------------------")
for word in text.split():
    print(word, "->", pos_tag(word))
```

10. Transformation-Based Tagging

```python
import nltk
from nltk import word_tokenize

nltk.download('punkt_tab', quiet=True)

sentence = "The dog runs fast"
words = word_tokenize(sentence)

tags = []
for word in words:
    if word in ["The", "the"]:
        tags.append((word, "DT"))
    elif word in ["dog", "cat"]:
        tags.append((word, "NN"))
    elif word.endswith("s"):
        tags.append((word, "VBZ"))
    else:
        tags.append((word, "RB"))

print("Tagged Words:", tags)
```

11. Top-Down Parser for Context-Free Grammar

```python
import nltk

grammar = nltk.CFG.fromstring("""
S -> NP VP
NP -> Det N
VP -> V NP
Det -> 'the'
N -> 'cat' | 'mouse'
V -> 'chased'
""")

parser = nltk.TopDownChartParser(grammar)
sentence = "the cat chased the mouse".split()

for tree in parser.parse(sentence):
    print(tree)
```

12. Earley Parser for Context-Free Grammar

```python
import nltk

grammar = nltk.CFG.fromstring("""
S -> NP VP
NP -> 'John' | 'Mary'
VP -> V NP
V -> 'likes'
""")

parser = nltk.EarleyChartParser(grammar)
sentence = "John likes Mary".split()

for tree in parser.parse(sentence):
    print(tree)
```

13. Generate a Parse Tree Using CFG

```python
import nltk

grammar = nltk.CFG.fromstring("""
S -> NP VP
NP -> Det N
VP -> V NP
Det -> 'the'
N -> 'boy' | 'apple'
V -> 'eats'
""")

parser = nltk.ChartParser(grammar)
sentence = "the boy eats the apple".split()

for tree in parser.parse(sentence):
    tree.pretty_print()
```

14. Sentence Agreement Checking Using CFG

```python
import nltk

grammar = nltk.CFG.fromstring("""
S -> SING | PLUR
SING -> 'He' 'runs'
PLUR -> 'They' 'run'
""")

parser = nltk.ChartParser(grammar)

sentences = ["He runs", "They run", "He run"]

for s in sentences:
    if list(parser.parse(s.split())):
        print(s, "-> Correct Agreement")
    else:
        print(s, "-> Incorrect Agreement")
```

15. Probabilistic Context-Free Grammar (PCFG)

```python
import nltk
from nltk import PCFG

grammar = PCFG.fromstring("""
S -> NP VP [1.0]
NP -> 'John' [0.5] | 'Mary' [0.5]
VP -> V NP [1.0]
V -> 'likes' [1.0]
""")

parser = nltk.ViterbiParser(grammar)
sentence = "John likes Mary".split()

for tree in parser.parse(sentence):
    print(tree)
    print("Probability:", tree.prob())
```

20. Information Retrieval Using TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

docs = [
    "Machine learning is interesting",
    "Python is used for programming",
    "Machine learning uses Python"
]

query = ["machine learning"]

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(docs + query)

scores = cosine_similarity(X[-1], X[:-1])

for i, score in enumerate(scores[0]):
    print("Document", i+1, ":", round(score, 3))
```

21. Syntax-Driven Semantic Analysis

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')

sentence = "The smart student reads books"

words = nltk.word_tokenize(sentence)
tags = nltk.pos_tag(words)

grammar = "NP: {<DT>?<JJ>*<NN.*>+}"
parser = nltk.RegexpParser(grammar)

result = parser.parse(tags)
print(result)

for subtree in result.subtrees():
    if subtree.label() == 'NP':
        print("Noun Phrase:", " ".join(w for w, t in subtree.leaves()))
```

23. Text Coherence Evaluation

```python
text = """
Python is a programming language.
Python is easy to learn.
Programming helps students.
"""

sentences = text.strip().split("\n")

for i in range(len(sentences)-1):
    w1 = set(sentences[i].lower().split())
    w2 = set(sentences[i+1].lower().split())

    score = len(w1 & w2) / len(w1 | w2)
    print("Coherence Score:", round(score, 2))
```

24. Dialog Act Recognition

```python
def dialog_act(sentence):
    s = sentence.lower()

    if s.startswith(("hi", "hello")):
        return "Greeting"
    elif s.endswith("?"):
        return "Question"
    elif "thank" in s:
        return "Thanking"
    elif s.startswith(("bye", "goodbye")):
        return "Goodbye"
    else:
        return "Statement"

dialog = [
    "Hello",
    "How are you?",
    "Thank you",
    "Goodbye"
]

for sentence in dialog:
    print(sentence, "->", dialog_act(sentence))
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# Load the tokenizer and model directly to bypass deprecated pipeline tasks
model_name = "Helsinki-NLP/opus-mt-en-fr"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

text = "How are you?"

# Tokenize and generate translation
inputs = tokenizer(text, return_tensors="pt")
translated_tokens = model.generate(**inputs)
result = tokenizer.decode(translated_tokens[0], skip_special_tokens=True)

print("English:", text)
print("French:", result)
```
